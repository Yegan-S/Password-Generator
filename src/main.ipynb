```python
from abc import ABC, abstractmethod


class PasswordGenerator(ABC):
    @abstractmethod
    def generate(self):
        pass
```

```python
class PinGenerator(PasswordGenerator):
    def __init__(self, lenght):
        self.lenght = lenght

    def generate(self):
        print("Hi")
```

```python
password_generator = PinGenerator(lenght=8)
password_generator.generate()
```
